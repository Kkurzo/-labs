### Завдання 19

```text

```

1. вертикально;

````python
A = np.array([
    [1, 2],
    [3, 4]
])
B = np.array([
    [5, 6],
    [7, 8]
])
vertical = np.vstack((A, B))
horizontal = np.hstack((A, B))

print("Вертикально:")
print(vertical)

print("Горизонтально:")
print(horizontal)

````
